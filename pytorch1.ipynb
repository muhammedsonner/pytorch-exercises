```python
x = torch.ones(5, requires_grad=True) # default false geliyor. gradyan iniş te geri yayılımda falan kullanabilmek için bu argümanı True yapmalıyız.
x
```

```python
x = torch.rand(3, requires_grad=True)
x
```

```python
y = x + 2
print(y) # işlem toplama olduğu için AddBackward
```

```python
z = y*y*2
z # işlem çarpma olduğu için MulBackward
```

```python
z = z.mean()
z # işlem ortalama olduğu için MeanBackward
```

```python
z.backward() # dz/dx
print(x.grad)
```

z skaler olunca argüman kullanmıyoruz. vektör olunca aşağıdaki işlemleri yapıyoruz

```python
x = torch.rand(3, requires_grad=True)
x
```

```python
y = x + 2
print(y) # işlem toplama olduğu için AddBackward
```

```python
z = y*y*2
z # işlem çarpma olduğu için MulBackward
```

```python
v = torch.tensor([0.1, 1.0, 0.01], dtype=torch.float32)
z.backward(v) # dz/dx
print(x.grad)

# geri yayılım yapılacak tensor skaler tek bir veriyse sıkıntı yok ama birden fazla değer içeriyorsa backward işlemi için bir vektör vermeliyiz
```

dummy operation

```python
weights = torch.ones(4, requires_grad=True)
```

```python
for epoch in range(3):
    model_output = (weights*3).sum()
    model_output.backward()
    print(weights.grad)
    weights.grad.zero_()
```

```python
x = torch.tensor(1.0)
y = torch.tensor(2.0)

w = torch.tensor(1.0, requires_grad=True)
```

```python
# forward pass ve hata hesaplama
y_pred = (x*w)
print("y_tahmin: ",y_pred)
s = (y_pred - y)
l = s**2
print("hata: ", l)
```

```python
# backward pass
l.backward()
w.grad
```
